# Evaluate the salary prediction models
Compare test results and inspect Linear Regression features.

```python
import joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
data_splits = joblib.load("data_splits.pkl")
linear_model = joblib.load("linear_model.pkl")
rf_model = joblib.load("rf_model.pkl")
feature_columns = joblib.load("feature_columns.pkl")
X_test_scaled = data_splits["X_test_scaled"]
y_test = data_splits["y_test"]
linear_predictions = linear_model.predict(X_test_scaled)
rf_predictions = rf_model.predict(X_test_scaled)
linear_mae = mean_absolute_error(y_test, linear_predictions)
linear_rmse = np.sqrt(mean_squared_error(y_test, linear_predictions))
linear_r2 = r2_score(y_test, linear_predictions)
rf_mae = mean_absolute_error(y_test, rf_predictions)
rf_rmse = np.sqrt(mean_squared_error(y_test, rf_predictions))
rf_r2 = r2_score(y_test, rf_predictions)
print("--- Model Results ---")
print("Linear Regression MAE: ₹{:,.2f}".format(linear_mae))
print("Linear Regression RMSE: ₹{:,.2f}".format(linear_rmse))
print("Linear Regression R2: {:.4f}".format(linear_r2))
print("Random Forest MAE: ₹{:,.2f}".format(rf_mae))
print("Random Forest RMSE: ₹{:,.2f}".format(rf_rmse))
print("Random Forest R2: {:.4f}".format(rf_r2))
```

```python
comparison = pd.DataFrame({
    "Model": ["Linear Regression", "Random Forest"],
    "MAE (₹)": [linear_mae, rf_mae],
    "RMSE (₹)": [linear_rmse, rf_rmse],
    "R2": [linear_r2, rf_r2]
})
comparison
```

```python
coefficients = pd.Series(linear_model.coef_, index=feature_columns)
print("--- Top 10 Features by Absolute Coefficient ---")
print(coefficients.abs().sort_values(ascending=False).head(10))
```

```python
plt.scatter(y_test, linear_predictions, alpha=0.6)
plot_min = min(y_test.min(), linear_predictions.min())
plot_max = max(y_test.max(), linear_predictions.max())
plt.plot([plot_min, plot_max], [plot_min, plot_max], color="red")
plt.xlabel("Actual Salary (₹)")
plt.ylabel("Predicted Salary (₹)")
plt.title("Linear Regression: Actual vs Predicted Salary")
plt.show()
```
