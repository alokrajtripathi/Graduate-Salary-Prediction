# Predict salaries for students
Use the saved model for a test student and one example new student.

```python
import joblib
import pandas as pd
```

```python
scaler = joblib.load("scaler.pkl")
feature_columns = joblib.load("feature_columns.pkl")
linear_model = joblib.load("linear_model.pkl")
data_splits = joblib.load("data_splits.pkl")
X_test_raw = data_splits["X_test_raw"]
y_test = data_splits["y_test"]
first_test_row = X_test_raw.iloc[0:1]
first_test_scaled = scaler.transform(first_test_row)
test_prediction = linear_model.predict(first_test_scaled)[0]
test_actual = y_test.iloc[0]
print("--- Part A: First Test Student ---")
print("Predicted salary: ₹{:,.2f}".format(test_prediction))
print("Actual salary: ₹{:,.2f}".format(test_actual))
```

```python
new_student = {
    "Gender": "m",
    "10percentage": 80,
    "12percentage": 78,
    "CollegeTier": 2,
    "Degree": "B.Tech/B.E.",
    "Specialization": "computer science & engineering",
    "collegeGPA": 7.5,
    "English": 500,
    "Logical": 600,
    "Quant": 550,
    "Domain": 0.6
}
new_student_df = pd.DataFrame([new_student])
new_student_encoded = pd.get_dummies(new_student_df, drop_first=True)
# Reindex adds missing training columns and orders them exactly as during training.
new_student_encoded = new_student_encoded.reindex(columns=feature_columns, fill_value=0)
new_student_scaled = scaler.transform(new_student_encoded)
new_student_prediction = linear_model.predict(new_student_scaled)[0]
print("--- Part B: Example New Student ---")
print("Predicted salary: ₹{:,.2f}".format(new_student_prediction))
```
